Pertanyaan:
1. Tanpa Matplotlib (menggunakan cv2.imshow bawaan OpenCV) maka akan membuka jendela baru atau menghasilkan error jika dijalankan di lingkungan cloud seperti Google Colab, sehingga memerlukan fungsi khusus seperti cv2_imshow. Berbeda dengan menggunakan Matplotlib (plt.imshow) maka gambar dirender langsung di dalam notebook sebagai bagian dari plot atau canvas inline, serta mendukung penambahan elemen plot seperti sumbu koordinat, judul, dan pengaturan colormap (cmap).

2. Tipe data int32 mengalokasikan memori dua kali lebih besar (4 byte per elemen) dibandingkan int16 (2 byte per elemen). Dan juga ntuk membuat citra hitam (black image atau zeros), umumnya menggunakan tipe data standar uint8 (0 hingga 255). Penggunaan int16 atau int32 tetap dapat menyimpan angka nol, namun tidak efisien untuk citra standar dan dapat menyebabkan masalah rentang nilai atau error pada fungsi OpenCV tertentu yang mengharapkan format unsigned integer 8-bit.

3. Berfungsi untuk mengimpor fungsi cv2_imshow khusus yang disediakan oleh Google Colab agar fungsi bawaan OpenCV (cv2.imshow) dapat berjalan dengan lancar di lingkungan browser Google Colab tanpa mengalami crash atau error GUI.
