```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
```

```python
img = cv2.imread("/content/drive/MyDrive/PCVK/KTM.png")
plt.imshow(img)
```

```python
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)) #konfersi BGR ke RGB
plt.show()
```

```python
img_gray = cv2.imread("/content/drive/MyDrive/PCVK/KTM.png", cv2.IMREAD_GRAYSCALE)

#tanpa cmap
plt.subplot(2,3,1)
plt.imshow(img_gray)
plt.title("Tanpa cmap")

#dengan cmap
plt.subplot(2,3,2)
plt.imshow(img_gray, cmap="gray")
plt.title("Dengan cmap gray")

#dengan cmap magma
plt.subplot(2,3,3)
plt.imshow(img_gray, cmap="magma")
plt.title("Dengan cmap magma")
```

```python
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) #konversi BGR ke RGB
imgRGBsmall = cv2.resize(imgRGB, (450,700))
plt.imshow(imgRGBsmall)
```

```python
imgFlip = cv2.flip(imgRGBsmall,1)
plt.imshow(imgFlip)
```

```python
fig = plt.figure(figsize=(10,10))
ax=fig.add_subplot(111)
ax.imshow(imgFlip)
```

```python
black_img = np.zeros(shape=(512,512,3), dtype=np.int16)
plt.imshow(black_img)
```

```python
#perhatikan koordinat titik2 pt1 dan pt2
cv2.rectangle(black_img, pt1=(384,0), pt2=(510,150), color=(0,255,0), thickness=10)
plt.imshow(black_img)
```

```python
cv2.rectangle(black_img, pt1=(200,200), pt2=(300,300), color=(0,0,255), thickness=10)
plt.imshow(black_img)
```

```python
cv2.circle(black_img, center=(100,100), radius=50, color=(255,0,0), thickness=10)
plt.imshow(black_img)
```

```python
cv2.circle(black_img, center=(400,400), radius=50, color=(255,0,0), thickness=-1)
plt.imshow(black_img)
```

```python
cv2.line(black_img, pt1=(0,0), pt2=(510,510), color=(255,255,255), thickness=5)
plt.imshow(black_img)
```

```python
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(black_img, text='Hello', org=(10,500), fontFace=font, fontScale=4, color=(255,255,0), thickness=2, lineType=cv2.LINE_AA)
plt.imshow(black_img)
```

```python
black_img2 = np.zeros(shape=(512,512,3), dtype=np.int32)
plt.imshow(black_img2)
```

```python
vertices = np.array([[100,300],[200,200],[400,300],[200,400]], dtype=np.int32)
vertices
```

```python
pts = vertices.reshape((-1, 1, 2))
pts
```

```python
cv2.polylines(black_img2, [pts], isClosed=True, color=(255,0,0), thickness=5)
plt.imshow(black_img2)
```
